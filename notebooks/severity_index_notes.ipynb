# Severity Index Experiments

Report: Definition and Application of the Disease Severity Index in Plant Pathology and Paddy Cultivation
## 1. Definition of the Disease Severity Index (DSI)

The Disease Severity Index (DSI), also referred to as the infection rating or severity index, is a synthetic quantitative estimate used to describe the intensity of a disease within a plant population
- While "disease severity" generally refers to the proportion of a sampling unit (such as a leaf or fruit) showing symptoms, the DSI provides a standardized way to summarize multiple observations
- Pathologists historically calculate the DSI in two primary ways:
Ratio-Based: It is defined as the ratio of the total area of lesions to the total infected leaf area
- Score-Based: When using ordinal rating scales (e.g., 0 to 6), the DSI is expressed as a weighted average: DI=100×mean of disease scores/maximum disease score
- This index was first proposed by McKinney in 1923 to quantify "degrees of infection" on wheat and has since become a standard metric for determining host resistance and the impact of disease on yield

## 2. Usage Across Different Fields
The measurement of disease severity is vital in several scientific and industrial sectors:
Agriculture and Plant Breeding: DSI is used to monitor epidemics, forecast yield loss, and evaluate the efficacy of chemical treatments like fungicides
- In breeding, it is essential for "phenotyping," or assessing which plant varieties are most resistant to specific pathogens
- Forestry: Severity estimation is applied at larger scales to assess defoliation and overall vegetative cover in tree stands
- Medicine (Comparative Usage): Innovative agricultural studies have adapted the ABCD rules (Asymmetry, Border irregularity, Colour variance, and Diameter) from human skin-cancer detection to classify the severity of plant lesions, as both fields rely on assessing visually strikes symptoms on a surface
- Food Quality and Safety: Hyperspectral Imaging (HSI), a tool for measuring severity, is also used to monitor fruit quality, detect fecal contamination on animal carcasses, and create land-use maps

## 3. Approaches to Detect Severity in Paddy Leaves
In the context of paddy (rice) leaves, detecting whether symptoms are due to a biotic disease or a nutrient deficiency is critical, as both can manifest as similar spectral responses or foliar discolorations

- A. Hyperspectral Imaging (HSI) HSI is one of the most powerful tools for distinguishing between diseases and deficiencies. For example, HSI has been used to determine brown spot disease severity in rice by analyzing the 350–2500 nm range of the electromagnetic spectrum
. Research indicates that specific wavelengths can differentiate biotic stress (like yellow rust) from abiotic stress (like nitrogen deficiency) in cereal canopies with an accuracy of 95–100% using neural net classifiers
.
- B. Machine Learning (ML) and Deep Learning (DL)
Traditional ML: Well-known methods like Support Vector Machines (SVM) and Random Forests (RF) are applied to hyperspectral data to estimate rice disease severity
.
Salient Object Detection (SOD): Lightweight DL models (e.g., MEANet and SeaNet) are used to segment visually striking objects—the disease lesions—from the rest of the paddy leaf
. This allows for a precise calculation of the area-based DSI even in field conditions with complex backgrounds
.
- C. Visual and Digital Image Analysis Digital photography in the visible spectrum (RGB) is frequently used to identify rice diseases such as bacterial leaf blight, brown spot, and leaf smut
. While visual estimation by human raters remains the most common method, it is prone to overestimating severity at low levels (under 10%) and can be influenced by "knots," where raters prefer reporting values in 5% or 10% increments
. Computer-aided systems like ASSESS now automate this process by thresholding pixels to distinguish healthy green tissue from necrotic or chlorotic areas
.

1. Automatic Pixel-Level Annotation (WSSS)
Because manual pixel-level annotation is laborious and subjective, you can use Weakly Supervised Semantic Segmentation (WSSS) to generate ground truth labels computationally
.
GradCAM and Vision Transformers (ViT): You can train a Vision Transformer as a feature extractor to recognize the labeled deficiencies
. By applying Gradient-weighted Class Activation Mapping (GradCAM), the model generates a heatmap (saliency map) that highlights the specific pixels contributing to the "deficiency" label
.
Masking: These heatmaps are then normalized and subjected to thresholding (commonly at values like 0.5 or 0.7) to create binary masks, which act as "pseudo-labels" for the symptomatic regions
.
2. Symptom and Leaf Segmentation
Once you have pseudo-labels, you must isolate the symptomatic areas and the leaf itself to quantify severity.
Salient Object Detection (SOD): You can train lightweight deep learning models like MEANet or SeaNet on your pseudo-labels
. These models are designed to segment "visually striking" objects—the lesions or discolorations—even against complex backgrounds
.
Leaf Area Calculation: To calculate the total area of the sampling unit, you must also segment the leaf
. For plain backgrounds, the marker-watershed method is a low-cost computational choice
. For complex field images, a separate SOD model should be trained specifically for leaf segmentation
.
3. Quantitative Severity Indexing
With the segmented masks, the Disease Severity Index (DSI)—which applies equally to nutrient deficiency symptoms—is calculated as the ratio of the symptomatic area to the total leaf area
.
Ratio Scale: This provides an objective, continuous quantitative value between 0 and 100%
.
Feature Extraction (ABCD & GLCM): Beyond simple area, you can extract high-level statistical features to describe the deficiency's progression:
ABCD Rules: Adapt the medical ABCD rules to calculate Asymmetry, Border irregularity, Colour variance, and Diameter of the symptomatic spots
.
Texture (GLCM): Use the Grey Level Co-occurrence Matrix to extract features like contrast, correlation, energy, and homogeneity to characterize the texture of the deficiency symptoms
.
4. Severity Level Classification
Finally, you can use these extracted features to classify the images into severity levels (e.g., Very Low, Low, High, Very High)
.
Machine Learning: Statistical features (ABCD + GLCM) can be fed into traditional classifiers like Support Vector Machines (SVM) or Random Forests, which are computationally efficient and can outperform deep learning baselines when data is limited
.
Feature Fusion: The most accurate approach involves a Feature Fusion Network, where deep visual features from the image are concatenated with the 30-dimensional statistical feature vector (DSI, ABCD, and GLCM) before the final classification layer
.
Performance and Accuracy
This computational approach is highly effective. For instance, the MEANet model using this pipeline achieved over 82% detection accuracy in severity classification tasks on leaf datasets
. Furthermore, fusing statistical features with visual data has been shown to improve the F1-score by approximately 10% over using visual data alone
.
